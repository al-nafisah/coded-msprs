# Symbol alphabet

Every value the modulator can emit at $L_0=3$, with how often each occurs over the trellis. The two families split the symbol energy differently, so they reach different level sets.

```python
import subprocess, pathlib, collections
import numpy as np
import matplotlib.pyplot as plt

ROOT = pathlib.Path.cwd().parent
MSPRS = ROOT / "build" / "bin" / "msprs"
FIGURES = ROOT / "figures"
FIGURES.mkdir(exist_ok=True)

plt.rcParams.update({
    "font.family": "sans-serif", "font.size": 8, "axes.labelsize": 8,
    "axes.titlesize": 8, "legend.fontsize": 7, "xtick.labelsize": 7,
    "ytick.labelsize": 7, "axes.linewidth": 0.6, "axes.grid": True,
    "grid.alpha": 0.3, "grid.linestyle": ":", "figure.dpi": 110,
    "savefig.dpi": 200, "savefig.bbox": "tight", "pdf.fonttype": 42,
})

def alphabet(L0, family):
    out = subprocess.run([str(MSPRS), "--mode", "alphabet", "--L0", str(L0),
                          "--family", family], capture_output=True, text=True,
                         check=True).stdout
    vals = [float(l.split()[3]) for l in out.splitlines() if not l.startswith("#")]
    return collections.Counter(round(v, 9) for v in vals)
```

```python
L0 = 3
fig, axes = plt.subplots(1, 2, figsize=(7.4, 3.0), sharey=True)
for ax, fam in zip(axes, ("balanced", "unbalanced")):
    c = alphabet(L0, fam)
    lv = sorted(c)
    st = ax.stem(lv, [c[v] for v in lv], basefmt=" ")
    st.markerline.set(color="#000000", markersize=4)
    st.stemlines.set(color="#000000", linewidth=1.0)
    ax.set(xlabel="symbol value", title=f"{fam}, {len(lv)} levels",
           ylim=(0, 2.6), yticks=[0, 1, 2])

axes[0].set_ylabel("multiplicity")
fig.tight_layout()
for ext in ("pdf", "png"):
    fig.savefig(FIGURES / f"constellations_overview.{ext}")

for fam in ("balanced", "unbalanced"):
    c = alphabet(L0, fam)
    print(f"{fam:11s} {len(c)} levels  {sorted(round(v,4) for v in c)}")
```
